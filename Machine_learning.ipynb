```python
def celsius_to_fahrenheit(celsius):
  return (celsius * 9 / 5) + 32

# Define threshold outside the function for use in the main loop
threshold_c = 25
print("Temperature converter and threshold checker")
print(f"Reference threshold: {threshold_c}°C\n")

while True:
  user_input = input("Enter temperature in °C (or type 'quit' to exit): ")
  if user_input.lower() == "quit":
    print("Exiting program. Goodbye!")
    break
  try:
    celsius = float(user_input)
    fahrenheit = celsius_to_fahrenheit(celsius)
    print(f"> {celsius}°C is equal to {fahrenheit:.2f}°F")

    if celsius > threshold_c:
      print(f"> Status: Above the {threshold_c}°C threshold.\n")
    elif celsius < threshold_c:
      print(f"> Status: Below the {threshold_c}°C threshold.\n")
    else:
      print(f"> Status: Exactly at the {threshold_c}°C threshold.\n")

  except ValueError:
    print("Invalid input. Please enter a valid number or type 'quit'.\n")
```
